## 1. Abstract Base Classes (ABC)
Le classi astratte forzano le sottoclassi a implementare determinati metodi. Non possono essere istanziate direttamente. In Python, si utilizza il modulo `abc`.

```python
from abc import ABC, abstractmethod

class Veicolo(ABC):
    @abstractmethod
    def avvia_motore(self):
        pass

class Auto(Veicolo):
    def avvia_motore(self):
        return "Vrum vrum! Il motore dell'auto è acceso."

mia_auto = Auto()
print(mia_auto.avvia_motore())
```

## 2. Ereditarietà Multipla e MRO (Method Resolution Order)
In Python, una classe può ereditare da più classi genitore. Il **Method Resolution Order (MRO)** definisce l'ordine con cui Python cerca i metodi e gli attributi (algoritmo C3 linearization).

```python
class A:
    def saluta(self):
        return "Ciao da A"

class B(A):
    def saluta(self):
        return "Ciao da B"

class C(A):
    def saluta(self):
        return "Ciao da C"

class D(B, C):
    pass

d = D()
print(d.saluta())  # B viene prima di C
print("MRO di D:", D.mro())
```

## 3. Class Methods e Static Methods
- `@classmethod`: Prende la classe (`cls`) come primo argomento. Utile per costruttori alternativi.
- `@staticmethod`: Non prende argomenti impliciti. Funziona come una normale funzione, ma organizzata logicamente nella classe.

```python
class Data:
    def __init__(self, giorno, mese, anno):
        self.giorno = giorno
        self.mese = mese
        self.anno = anno

    @classmethod
    def da_stringa(cls, data_str):
        g, m, a = map(int, data_str.split('-'))
        return cls(g, m, a)

    @staticmethod
    def e_bisestile(anno):
        return anno % 4 == 0 and (anno % 100 != 0 or anno % 400 == 0)

d = Data.da_stringa("22-10-2026")
print(f"Data creata: {d.giorno}/{d.mese}/{d.anno}")
print(f"Il 2026 è bisestile? {Data.e_bisestile(2026)}")
```

## 4. Magic Methods (Dunder Methods)
I metodi speciali permettono di personalizzare il comportamento degli oggetti Python (operazioni matematiche, stampa, ecc.).

```python
class Vettore:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, altro):
        return Vettore(self.x + altro.x, self.y + altro.y)

    def __str__(self):
        return f"Vettore({self.x}, {self.y})"

v1 = Vettore(2, 3)
v2 = Vettore(4, 1)
print(f"{v1} + {v2} = {v1 + v2}")
```

## 5. Context Managers
Gestiscono le risorse definendo le operazioni da fare all'ingresso (`__enter__`) e all'uscita (`__exit__`) di un blocco `with`.

```python
import time

class Timer:
    def __enter__(self):
        self.start = time.time()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.end = time.time()
        print(f"Tempo trascorso: {self.end - self.start:.4f} secondi")

with Timer():
    sum(i * i for i in range(1000000))
```

## 6. Descriptors
I descrittori permettono di definire il comportamento quando si accede o si modifica un attributo, implementando `__get__`, `__set__` o `__delete__`.

```python
class TemperaturaPositiva:
    def __get__(self, instance, owner):
        return instance._temperatura

    def __set__(self, instance, valore):
        if valore < -273.15:
            raise ValueError("Troppo freddo!")
        instance._temperatura = valore

class Stazione:
    temperatura = TemperaturaPositiva()

s = Stazione()
s.temperatura = 25
print(s.temperatura)
```

## 7. Metaclasses
Le metaclassi definiscono *come* una classe viene creata. Puoi usarle per modificare la struttura della classe prima ancora che venga istanziata.

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Database(metaclass=SingletonMeta):
    pass

db1 = Database()
db2 = Database()
print("Sono la stessa istanza?", db1 is db2)
```
